```python
competition_url = BASE_URL + "competitions/PL/matches"

# Making the request (asking the API for data)
response = requests.get(competition_url, headers = headers)

#Checking if it worked
if response.status_code  == 200:
    print("Data fetched successfully")
    data = response.json()

else:
    print("Error: {response.status_code}")
    print("Check your API key")
```

```python
#Lets see whats in the data
print("Keys in our data:", data.keys())
print("\n" + "="*50 + "\n")

#Get the matches
matches = data ['matches']
print(f"Total matches fetched: {len(matches)}")
print("\n" + "="*50 + "\n")

print("Here's what ONE match looks like:")
print (matches[0])

```

```python
#Lets create an empty list to store information
match_data = []

#To loop through each match adn extract the info we want
for match in matches:
    match_info = {
        'date': match['utcDate'][:10], #Just the date, not time
        'home_team': match['homeTeam']['name'],
        'away_team': match['awayTeam']['name'],
        'home_score': match['score']['fullTime']['home'],
        'away_score': match['score']['fullTime']['away'],
        'status': match['status'],
        'matchday': match['matchday']}
    match_data.append(match_info)

#Now lets convert the data to a dataframe
df = pd.DataFrame(match_data)

print("Data organized into a table!")
print(f"\nDataFrame shape: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"\nFirst 5 matches:")
df.head()
```

```python
#Lets check how many matches are there in each status
print("Match statuses:")
print(df['status'].value_counts())
print("\n" + "="*50 + "\n")

#Lets only keep the games that are finished:
df_finished = df[df['status'] == "FINISHED"].copy()

print(f"Total finished matches: {len(df_finished)}")
print ("\n" + "=" *50 + "\n")

#Now, lets remove the rows that have missing scores:
df_finished = df_finished.dropna(subset=['home_score', 'away_score'])

#Since scores are neever floats, lets convert them to integers
df_finished['home_score'] = df_finished['home_score'].astype(int)
df_finished['away_score'] = df_finished['away_score'].astype(int)

print("Data cleaned!")
print("\nFirst 5 finished matches:")
df_finished.head()
```

```python
#Lets calculate now, the total goals for each match
df_finished["total_goals"] = df_finished["home_score"] + df_finished["away_score"]

#Then we calculate, the goal difference for home team
df_finished["goal_difference"] = df_finished["home_score"] - df_finished["away_score"]

#This function will evaluate if the home_team won or lost
def get_result(row):
    if row['home_score'] > row['away_score']:
        return 'Home Win'
    elif row['home_score'] < row['away_score']:
        return 'Away Win'
    else:
        return 'Draw'

#Apply the function 
df_finished['result'] = df_finished.apply(get_result, axis=1)

#Create a new table with added columns
print("New columns created!")
print("\nDataFrame with new columns")
df_finished.head(10)
                                         
```

```python
print("="*50)
print("PREMIER LEAGUE STATISTICS")
print("="*50)

print(f"\nTotal Matches analyzed: {len(df_finished)}")
print(f"\nAverage Goals per match: {df_finished['total_goals'].mean():.2f}")
print(f"\nHome team average score: {df_finished['home_score'].mean():.2f}")
print(f"\nAway team average score: {df_finished['away_score'].mean():.2f}")

print("\n"+ "="*50)
print("MATCH RESULTS BREAKDOWN")
print("="*50)
print(df_finished['result'].value_counts())

print("\n"+"="*50)
print("HIGHEST SCORING MATCHES")
print('='*50)
print(df_finished.nlargest(5, 'total_goals')[['date', 'home_team', 'away_team', 'home_score', 'away_score', 'total_goals']])
```
