```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier 
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix 
import warnings
warnings.filterwarnings('ignore')

#setting style for better looking plots
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12,6)

print("All Libraries imported successfully")
```

```python
competition_url = BASE_URL + "competitions/PL/matches"

# Making the request (asking the API for data)
response = requests.get(competition_url, headers = headers)

#Checking if it worked
if response.status_code  == 200:
    print("Data fetched successfully")
    data = response.json()

else:
    print("Error: {response.status_code}")
    print("Check your API key")
```

```python
#Lets see whats in the data
print("Keys in our data:", data.keys())
print("\n" + "="*50 + "\n")

#Get the matches
matches = data ['matches']
print(f"Total matches fetched: {len(matches)}")
print("\n" + "="*50 + "\n")

print("Here's what ONE match looks like:")
print (matches[0])

```

```python
#Lets create an empty list to store information
match_data = []

#To loop through each match adn extract the info we want
for match in matches:
    match_info = {
        'date': match['utcDate'][:10], #Just the date, not time
        'home_team': match['homeTeam']['name'],
        'away_team': match['awayTeam']['name'],
        'home_score': match['score']['fullTime']['home'],
        'away_score': match['score']['fullTime']['away'],
        'status': match['status'],
        'matchday': match['matchday']}
    match_data.append(match_info)

#Now lets convert the data to a dataframe
df = pd.DataFrame(match_data)

print("Data organized into a table!")
print(f"\nDataFrame shape: {df.shape[0]} rows, {df.shape[1]} columns")
print(f"\nFirst 5 matches:")
df.head()
```

```python
#Lets check how many matches are there in each status
print("Match statuses:")
print(df['status'].value_counts())
print("\n" + "="*50 + "\n")

#Lets only keep the games that are finished:
df_finished = df[df['status'] == "FINISHED"].copy()

print(f"Total finished matches: {len(df_finished)}")
print ("\n" + "=" *50 + "\n")

#Now, lets remove the rows that have missing scores:
df_finished = df_finished.dropna(subset=['home_score', 'away_score'])

#Since scores are neever floats, lets convert them to integers
df_finished['home_score'] = df_finished['home_score'].astype(int)
df_finished['away_score'] = df_finished['away_score'].astype(int)

print("Data cleaned!")
print("\nFirst 5 finished matches:")
df_finished.head()
```

```python
#Lets calculate now, the total goals for each match
df_finished["total_goals"] = df_finished["home_score"] + df_finished["away_score"]

#Then we calculate, the goal difference for home team
df_finished["goal_difference"] = df_finished["home_score"] - df_finished["away_score"]

#This function will evaluate if the home_team won or lost
def get_result(row):
    if row['home_score'] > row['away_score']:
        return 'Home Win'
    elif row['home_score'] < row['away_score']:
        return 'Away Win'
    else:
        return 'Draw'

#Apply the function 
df_finished['result'] = df_finished.apply(get_result, axis=1)

#Create a new table with added columns
print("New columns created!")
print("\nDataFrame with new columns")
df_finished.head(10)
                                         
```

<br>

```python
#We need to give each team a performance rating based on their history so let's calculate some statistics for each team

team_stats = {}

for team in df_finished['home_team'].unique():
    #Games as home team
    home_games = df_finished[df_finished['home_team'] == team]
    #Games as away team
    away_games = df_finished[df_finished['away_team'] == team]


    #Now lets calculate the averages
    team_stats[team] = {
        'avg_goals_scored_home': home_games['home_score'].mean() if len(home_games) > 0 else 0,
        'avg_goals_conceded_home': home_games['away_score'].mean() if len(home_games) > 0 else 0, 
        'avg_goals_scored_away': away_games['away_score'].mean() if len(away_games) > 0 else 0,
        'avg_goals_conceded_away': away_games['home_score'].mean() if len(away_games) > 0 else 0,
    }

print(f"\nExample: Let's look at one team's stats:")
first_team = list(team_stats.keys())[0]
print(f"\n{first_team}:")
for stat, value in team_stats[first_team].items():
    print(f" {stat}: {value:.2f}")
    
```

```python
#Now let's add these features to our matches
feature_list = []

for idx, row in df_finished.iterrows():
    home = row['home_team']
    away = row['away_team']

    features = {
    'home_avg_scored': team_stats[home]['avg_goals_scored_home'],
    'home_avg_conceded': team_stats[home]['avg_goals_conceded_home'],
    'away_avg_scored': team_stats[away]['avg_goals_scored_away'],
    'away_avg_conceded': team_stats[away]['avg_goals_conceded_away'],
    'result': row['result']
    }
    feature_list.append(features)

#Lets create a new dataframe with features
ml_df = pd.DataFrame(feature_list)

print(f"\nDataset shape: {ml_df.shape}")
print("\nFirst 5 rows:")
ml_df.head()
```
